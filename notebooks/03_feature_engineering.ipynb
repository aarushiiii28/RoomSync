# Difference Features
The model doesn't care much about:

Student A studies 7 hours
Student B studies 2 hours

It cares about:

Difference = 5

```python
difference_features = [
    "sleep_time",
    "wake_time",

    # Work / focus
    "study_hours",
    "academic_seriousness",

    # Cleanliness
    "cleanliness_score",
    "organization_score",

    # Privacy / boundaries
    "privacy_preference",
    "boundary_importance",

    # Social
    "talkativeness",
    "friendship_expectation",
    "guest_frequency",

    # Lifestyle
    "gaming_hours",
    "music_volume",

    # Environment
    "noise_sleep_tolerance",

    # Personality / routine
    "schedule_consistency",
    "punctuality",
    "adaptability"
]

for feature in difference_features:
    col1 = f"{feature}_1"
    col2 = f"{feature}_2"

    if col1 in df.columns and col2 in df.columns:
        df[f"{feature}_diff"] = abs(
            df[col1] - df[col2]
        )

print("Difference features created.")
```

```python
print(df.shape)
print([col for col in df.columns if "_diff" in col])
```

# Similarity Features

```python

max_ranges = {
    # Work / focus
    "study_hours": 6,              # 1–7 scale
    "academic_seriousness": 4,     # 1–5 scale

    # Cleanliness
    "cleanliness_score": 4,
    "organization_score": 4,

    # Privacy / boundaries
    "privacy_preference": 4,
    "boundary_importance": 4,

    # Social
    "talkativeness": 4,
    "friendship_expectation": 4,
    "guest_frequency": 4,

    # Lifestyle
    "gaming_hours": 8,
    "music_volume": 4,

    # Environment
    "noise_sleep_tolerance": 4,

    # Routine / personality
    "schedule_consistency": 4,
    "punctuality": 4,
    "adaptability": 4,
}

for feature in difference_features:

    diff_col = f"{feature}_diff"
    similarity_col = f"{feature}_similarity"

    if diff_col not in df.columns:
        continue

    # Time features use circular similarity
    if feature in ["sleep_time", "wake_time"]:
        df[similarity_col] = (
            1 - (df[diff_col] / 12)
        ).clip(lower=0)

    else:
        max_diff = max_ranges.get(feature)

        if max_diff is not None:
            df[similarity_col] = (
                1 - (df[diff_col] / max_diff)
            ).clip(lower=0)

print("Similarity features created.")
```

# Aggregate Compatibility Features

```python

# Sleep / daily routine
df["sleep_compatibility"] = (
    df["sleep_time_similarity"]
    + df["wake_time_similarity"]
) / 2


# Work / academic discipline
df["work_compatibility"] = (
    df["study_hours_similarity"]
    + df["academic_seriousness_similarity"]
) / 2


# Cleanliness & organization
df["cleanliness_compatibility"] = (
    df["cleanliness_score_similarity"]
    + df["organization_score_similarity"]
) / 2


# Social compatibility
df["social_compatibility"] = (
    df["talkativeness_similarity"]
    + df["friendship_expectation_similarity"]
    + df["guest_frequency_similarity"]
) / 3


# Privacy & boundaries
df["privacy_compatibility"] = (
    df["privacy_preference_similarity"]
    + df["boundary_importance_similarity"]
) / 2


# Routine / discipline
df["routine_compatibility"] = (
    df["schedule_consistency_similarity"]
    + df["punctuality_similarity"]
    + df["adaptability_similarity"]
) / 3
```

# Archetype Features

```python
df["same_lifestyle_type"] = (

    df["lifestyle_type_1"]

    ==

    df["lifestyle_type_2"]

).astype(int)
```

# Composite Compatibility Score

```python
df["behavioral_alignment_score"] = (
    df["sleep_compatibility"]
    + df["work_compatibility"]
    + df["cleanliness_compatibility"]
    + df["social_compatibility"]
    + df["privacy_compatibility"]
    + df["routine_compatibility"]
) / 6
```
